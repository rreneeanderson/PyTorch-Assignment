## Part 5 — Matrix Multiplication
Create:
```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[7, 8],
              [9, 10],
              [11, 12]])
```
1.	Compute A @ B.
2.	Compute the same result using np.dot(A, B).
3.	Verify that both results are identical.
4.	Print the shape of A, B, and the result.

The notebook demonstrates the equivalence of .dot(), np.dot(), and @ for matrix multiplication.

````python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[7, 8],
              [9, 10],
              [11, 12]])

result1 = A @ B
result2 = np.dot(A, B)

print(result1)
print(result2)

print(np.array_equal(result1, result2))

print(A.shape)
print(B.shape)
print(result1.shape)
````

## Part 7 — Mini Application: Student Grades

Create a NumPy array containing the grades of 5 students in 4 assignments:

* Student 1: 85, 90, 78, 92
* Student 2: 70, 88, 91, 76
* Student 3: 95, 93, 89, 97
* Student 4: 60, 72, 68, 75
* Student 5: 88, 84, 90, 86

Using NumPy only:
1.	Calculate each student’s average.
2.	Find the highest grade on each assignment.
3.	Find the average grade for each assignment.
4.	Find the student with the highest overall average.
5.	Find all grades below 70.
6.	Replace grades below 70 with 70.
7.	Calculate the overall class average.

````python
grades = np.array([
    [85, 90, 78, 92],
    [70, 88, 91, 76],
    [95, 93, 89, 97],
    [60, 72, 68, 75],
    [88, 84, 90, 86]
])

print(np.mean(grades, axis=1))

print(np.max(grades, axis=0))

print(np.mean(grades, axis=0))

student_averages = np.mean(grades, axis=1)
print(np.argmax(student_averages) + 1)

print(grades[grades < 70])

grades[grades < 70] = 70
print(grades)
````

## Short Questions

Answer in your own words:
1.	What is the difference between shape, size, and ndim?
Shape tells the number of rows and columns, size tells the total number of elements, and ndim tells how many dimensions the array has.
2.	What is the difference between element-wise multiplication and matrix multiplication?
Element-wise multiplication multiplies values in the same positions. Matrix multiplication combines rows and columns to create a new matrix.
3.	What does axis=0 mean for a two-dimensional array?
axis=0 works down the rows, so it gives a result for each column.
4.	What does axis=1 mean?
axis=1 works across the columns, so it gives a result for each row.
5.	Why is NumPy useful for machine learning and deep learning?
NumPy makes it easier and faster to work with large amounts of numerical data and perform math on arrays and matrices.

## Challenge Problem

Create a random 5 × 5 matrix containing integers from 1 to 100.

Using NumPy only:
1.	Find the largest value.
2.	Find the smallest value.
3.	Calculate the mean.
4.	Calculate the standard deviation.
5.	Find all values greater than the mean.
6.	Replace all values below the mean with 0.
7.	Calculate the sum of each row and each column.

````python
matrix = np.random.randint(1, 101, (5, 5))

print(matrix)

print(np.max(matrix))
print(np.min(matrix))

mean = np.mean(matrix)
print(mean)

print(np.std(matrix))

print(matrix[matrix > mean])

matrix[matrix < mean] = 0
print(matrix)

print(np.sum(matrix, axis=1))
print(np.sum(matrix, axis=0))
````
