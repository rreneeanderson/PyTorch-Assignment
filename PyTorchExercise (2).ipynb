Part 9 — Automatic Differentiation

PyTorch provides automatic differentiation using autograd.

````python
x = torch.tensor([[1., 2.],
                  [3., 4.]], requires_grad=True)

# TODO: Calculate y = 3x² + 2
y = 3 * x**2 + 2
````

````python
# Then calculate:
z = sum(y)

# Use:
z = torch.sum(y)
z.backward()
# to calculate: dz/dx

# Print:

print(y)
print(z)
print(x.grad)

````

Questions
* What does requires_grad=True mean?
tells PyTorch to keep track of calculations so it can find gradients.
* What does backward() do?
calculates the gradients.
* What is stored in x.grad?
stores the gradient values for x
* Why is automatic differentiation important for neural networks?
Automatic differentiation is important because neural networks use gradients to update their weights during training
